```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv("light_spotify_dataset.csv")
data
```

```python
data.columns
```

```python
data.describe()
```

```python
data.dtypes
```

```python
data['Release Date'] = pd.to_datetime(data['Release Date'])
data.dtypes
```

```python
data.columns = data.columns.str.replace(" ", "_").str.replace("-", "_")
data
```

```python
data.drop_duplicates()
```

```python
data.isna().sum()
```

```python
data[data['song'].isna()]
```

```python
data = data.dropna(subset=['song'])
data.isna().sum()
```

```python
data['artist'].value_counts()
```

```python
data['Release_Date'] = data['Release_Date'].astype(str).str[-4:]
data
```

```python
import plotly.express as px
```

```python
data_grouped_by_date = data.groupby('Release_Date')['song'].nunique().reset_index()


fig = px.bar(
    data_grouped_by_date,
    x = 'Release_Date',
    y = 'song',
    template='presentation'
)

fig.show()
```

```python
data_grouped_by_date
```

```python
data_grouped_by_artists = data.groupby('artist')['Popularity'].mean().reset_index()
data_grouped_by_artists
```

```python
figure_2 = px.bar(
    data_grouped_by_artists,
    x = 'artist',
    y = 'Popularity',
    template = 'presentation'
)
figure_2.show()
```
